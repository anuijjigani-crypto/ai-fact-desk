# Exploratory Data Analysis — Fake and Real News Dataset
Run `python src/data_preprocessing.py` first so `data/processed/*.csv` exist.

```python
import sys, os
sys.path.append(os.path.abspath(".."))
import pandas as pd
import matplotlib.pyplot as plt
import config

train = pd.read_csv(config.TRAIN_CSV)
train.head()
```

```python
train['label'].value_counts().rename(index=config.ID2LABEL).plot(kind='bar', title='Class balance')
plt.show()
```

```python
train['word_count'] = train['content'].str.split().apply(len)
train.groupby('label')['word_count'].describe()
```

```python
train['word_count'].hist(bins=50)
plt.title('Article length distribution')
plt.xlabel('words')
plt.show()
```
